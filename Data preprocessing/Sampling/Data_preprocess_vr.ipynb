```python
import os
import random
import shutil
from tqdm import tqdm

# -----------------------------
# PATHS (EDIT THESE)
# -----------------------------

ANN_DIR = "/Users/gdsatishkumar/Downloads/annos_updated"
IMG_DIR = "/Users/gdsatishkumar/Downloads/images_filtered_all"

OUT_IMG_DIR = "/Users/gdsatishkumar/Downloads/sample_35k/images"
OUT_ANN_DIR = "/Users/gdsatishkumar/Downloads/sample_35k/annos"

SAMPLE_SIZE = 35000

# -----------------------------
# CREATE OUTPUT FOLDERS
# -----------------------------

os.makedirs(OUT_IMG_DIR, exist_ok=True)
os.makedirs(OUT_ANN_DIR, exist_ok=True)

# -----------------------------
# LIST ALL ANNOTATIONS
# -----------------------------

all_annos = [f for f in os.listdir(ANN_DIR) if f.endswith(".json")]

print("Total annotation files:", len(all_annos))

# -----------------------------
# RANDOM SAMPLE
# -----------------------------

sampled_annos = random.sample(all_annos, SAMPLE_SIZE)

# sort so both folders remain in same order
sampled_annos.sort()

print("Sampled:", len(sampled_annos))

# -----------------------------
# COPY FILES
# -----------------------------

missing_images = 0

for ann in tqdm(sampled_annos):

    ann_path = os.path.join(ANN_DIR, ann)
    img_name = ann.replace(".json", ".jpg")
    img_path = os.path.join(IMG_DIR, img_name)

    # copy annotation
    shutil.copy2(ann_path, os.path.join(OUT_ANN_DIR, ann))

    # copy corresponding image
    if os.path.exists(img_path):
        shutil.copy2(img_path, os.path.join(OUT_IMG_DIR, img_name))
    else:
        missing_images += 1

print("\nSampling completed.")
print("Missing images:", missing_images)
print("Images saved to:", OUT_IMG_DIR)
print("Annotations saved to:", OUT_ANN_DIR)
```
